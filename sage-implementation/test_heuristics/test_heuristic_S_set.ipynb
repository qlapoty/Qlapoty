```python
# Function taken from the sum of 2 square files

# Return a, b such that a^2 + b^2 = n if such a, b exist.
def SumOf2Sq(n):
    if n < 0:
        return None
    if n == 0:
        return 0, 0
    if n == 1:
        return 1, 0
    a, b = 1, 0
    for q in Primes()[:100]:
        e = 0
        while n % q == 0:
            n //= q
            e += 1
        a *= q**(e//2)
        b *= q**(e//2)
        if e % 2 == 1:
            if q % 4 == 3:
                return None
            else:
                s, t = cornacchia_smith(q)
                at = a*s - b*t
                bt = a*t + b*s
                a, b = at, bt
    if n % 4 == 1 and is_pseudoprime(n):
        s, t = cornacchia_smith(n)
        at = a*s - b*t
        bt = a*t + b*s
        a, b = at, bt
    elif n > 1:
        return None
    return a, b

# Return a, b such that a^2 + b^2 = q, where q is a prime.
def cornacchia_smith(q):
    if q == 2:
        a, b = 1, 1
    else:
        x = Mod(-1, q).sqrt()
        a, b, c = q, ZZ(x), floor(sqrt(q))
        while b > c:
            a, b = b, a % b
        t = q - b**2
        a, b = b, sqrt(t)

    # randomize output
    r = randint(0, 1)
    if r == 0:
        a, b = b, a
    r1 = randint(0, 1)
    r2 = randint(0, 1)
    return (-1)**r1*a, (-1)**r2*b



        
        



```

## Study of Assumption n°2 of the Qlapoty paper

In this jupyter notebook, we expermimentaly verify the assumption n°2 on the $S_i(a,b)$ set Qlapoty paper. Namely, we verify that, if $b - a > log(N)^2$, then there exists $n$ such that:
- $a \leq n \leq b$.
- $n = c^2 + d^2$.
- $\gcd(c, d) = 1$.
- $\gcd(n, 2N) = 1$.

```python
def construct_S(a,b):
    S = []
    for i in range(a, b):
        res = SumOf2Sq(i)
        if res == None:
            continue
        c, d = res
        if gcd(c, d) == 1:
            S.append(i)
    return S

proof.all(False)

N = randint(1, 2**64)
N_sqr = int(sqrt(N))
N_log = ZZ(N).nbits()
Bound = N_log**2

#Test 250 cases
for i in range(250):
    b = randint(Bound, N_sqr)
    a = b - Bound

    #print("a, b = " , a, b)

    assert(a < b)
    assert(b < N_sqr)

    S = construct_S(a, b)

    found_match = False
    for n in S:
        if gcd(2*N, n) == 1:
            assert(a <= n and n <= b)
            print(i , "Match = ", n)
            found_match = True
            break
    
    if(not found_match):
        print("No match", N, a ,b)
        
print("All test passed!")

        
```
